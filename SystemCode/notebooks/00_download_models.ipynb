# Download PropertyLens models from Hugging Face

Pulls the full **`artifacts/**`** tree from the model repo [`PropertyLens/propertylens-models`](https://huggingface.co/PropertyLens/propertylens-models/tree/main/artifacts) into the project **`data/`** tree (local path: `data/artifacts/...`, including `hybrid_xai/`, cluster bundles, etc.).

Requires **`HF_TOKEN`** in the repo-root `.env` (same as other HF notebooks).

The API loads artefacts from **`data/processed/`** by default (`PROPERTYLENS_ARTIFACTS_DIR` overrides). This notebook stores files under **`data/artifacts/`**; copy or symlink into `data/processed/` if you want the backend to use them without changing env.

## Configure paths and token

```python
import os
from pathlib import Path

from dotenv import load_dotenv
from huggingface_hub import snapshot_download

_cwd = Path.cwd().resolve()
REPO_ROOT = _cwd.parent if _cwd.name == "notebooks" else _cwd

load_dotenv(REPO_ROOT / ".env")
hf_token = os.getenv("HF_TOKEN")
if not hf_token:
    raise ValueError("HF_TOKEN not set — add it to .env at the repo root.")

REPO_ID = "PropertyLens/propertylens-models"
REPO_TYPE = "model"
# Files land under: <REPO_ROOT>/data/artifacts/...
LOCAL_DATA = REPO_ROOT / "data"
LOCAL_DATA.mkdir(parents=True, exist_ok=True)

print(f"Repo root: {REPO_ROOT}")
print(f"Download into: {LOCAL_DATA}")
```

## Download `artifacts/`

```python
path = snapshot_download(
    repo_id=REPO_ID,
    repo_type=REPO_TYPE,
    token=hf_token,
    local_dir=str(LOCAL_DATA),
    allow_patterns=["artifacts/**"],
)
print("Done.")
print(path)
```

## (Optional) List downloaded files

```python
target = LOCAL_DATA / "artifacts"
if target.exists():
    files = sorted(p for p in target.rglob("*") if p.is_file())
    print(f"{len(files)} file(s) under {target}:")
    for p in files:
        rel = p.relative_to(LOCAL_DATA)
        size_mb = p.stat().st_size / (1024 * 1024)
        print(f"  {rel}  ({size_mb:.2f} MB)")
else:
    print(f"Not found: {target}")
```
