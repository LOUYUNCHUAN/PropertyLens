## Code to pull a test dataset file from Hugging Face

### Before running the next cell, Please create a Hugging Face Token and add that to .env file

```python
import os
from dotenv import load_dotenv
from huggingface_hub import hf_hub_download
import pandas as pd

load_dotenv()

# Download the raw CSV file directly
path = hf_hub_download(
    repo_id="PropertyLens/Resealeflats",
    filename="Resale Flat Prices (Based on Registration Date), From Jan 2015 to Dec 2016.csv",
    repo_type="dataset",
    token=os.getenv("HF_TOKEN")
)

df = pd.read_csv(path)
print(df.shape)
print(df.head())
```

## Code to push a dataset to Hugging Face

```python
import os
from dotenv import load_dotenv
from huggingface_hub import HfApi

load_dotenv()

api = HfApi(token=os.getenv("HF_TOKEN"))

# Upload a single CSV
api.upload_file(
    # here movies_metadata.csv is your local file path
    path_or_fileobj="movies_metadata.csv",
    path_in_repo="raw/your_file.csv",
    repo_id="PropertyLens/Resealeflats",
    repo_type="dataset",
    commit_message="Add newd resale data CSV"
)

```
